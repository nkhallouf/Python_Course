## Loops
This notebook will just go through the basic topics in order:
    * For Loops

### For Loops
A for loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, a string, or range).

```python
# This code prints the numbers from 0 to 4 (inclusive) using a for loop that iterates over a range from 0 to n-1 (where n = 5). 
n = 5
for i in range(0, n):
    print(i)
```

###  While Loop
With the while loop we can execute a set of statements as long as a condition is true.

```python
i = 1
while i < 6:
  print(i)
  i += 1
#Note: remember to increment i, or else the loop will continue forever.    
```

### Nested Loops
Python programming language allows to use one loop inside another loop which is called nested loop. Following example illustrates the concept.

```python
# for every value of i, the loop over j starts again from the beginning
for i in range(1, 4):        # Outer loop (i) controls the rows
    for j in range(1, 4):    # Inner loop (j) runs completely for each iteration of the outer loop
        print(i, "x", j, "=", i * j)
    print("---")
```

### Loop Control Statements
Loop control statements change execution from their normal sequence. When execution leaves a scope, all automatic objects that were created in that scope are destroyed. Python supports the following control statements. 

```python
# continue → skips one iteration
for i in range(5):
    if i == 2:
        continue
    print(i)
```

```python
# pass → does nothing (used as placeholder)
for i in range(3):
    if i == 1:
        pass
    print(i)
```

```python
# break → exits the loop
for i in range(5):
    if i == 3:
        break
    print(i)
```
