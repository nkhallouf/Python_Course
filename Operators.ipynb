## Operators
This notebook will just go through the basic topics in order:


    * Arithmetic operators
    * Assignment operators

    * Logical operators
    * Identity operators
    * Membership operators
 


### Arithmetic operators
They are used with numeric values

```python
x = 15
y = 4

print(x + y)    # Addition
print(x - y)    # Subtraction
print(x * y)    # Multiplication
print(x / y)    # Division
print(x % y)    # Modulus
print(x ** y)   # Exponentiation
print(x // y)   # Floor division
```

### Assignment operators
Assignment operators are used to assign values to variables:

```python
x = 5  # =
x
```

```python
x += 3    # same as  x = x + 3
x
```

```python
x = 5
x -= 3    # same as x = x - 3
x
```

and so on for /= , %= , //= , **= , &= , |=,   ......

### Comparison Operators
Comparison operators are used to compare two values:

```python
1 > 2    # Greater than    
```

```python

'hi' != 'bye'  # Not equal
```

```python
'hi' == 'bye'  # Equal
```

```python
1 <= 4   # Less than or equal to
```

```python
1 >= 1   # Greater than or equal to
```

```python
1 < 2    # Less than
```

```python
1 == 1   # Equal
```

### Logic Operators
Logical operators are used to combine conditional statements:

```python
 # and  Returns True if both statements are true
(1 > 2) and (2 < 3)
```

```python
#  	Returns True if one of the statements is true
(1 == 2) or (2 == 3) or (4 == 4)
```

```python
# Reverse the result, returns False if the result is true
print(not(5 > 3 and 5 < 10))
```

### Identity Operators
Identity operators are used to compare the objects, not if they are equal, but if they are actually the same object, with the same memory location:

```python
# is  Returns True if both variables are the same object
x = ["Identity", "operators"]
y = ["Identity", "operators"]
z = x

print(x is z)
print(x is y) # be carefull they are not the same object
print(x == y)
```

```python
# The is not operator returns True if both variables do not point to the same object:
x = ["Identity", "operators"]
y = ["Identity", "operators"]
print(x is not y)
```

### Membership Operators
Membership operators are used to test if a sequence is presented in an object

```python
# in  Returns True if a sequence with the specified value is present in the object
List = ["I", "love", "Python"]
print("Python" in List)
```

```python
# not in Returns True if a sequence with the specified value is not present in the object
List = ["I", "love", "Python"]
print("Java" not in List)
```
