## Conditional Statements
This notebook will just go through the basic topics in order:
    * if statement
    * elif statement
    * else statement
    * Nested If Statements
    * pass statement

###  if statement
The if statement evaluates a condition (an expression that results in True or False). If the condition is true, the code block inside the if statement is executed. If the condition is false, the code block is skipped.

```python
# Checking if an age is Greater than or equal to 18:
age = 20
if age >= 18:
    print("You can vote.")
```

```python
# Short Hand If
age = 19
if age > 18: print("You to Vote.")
```

### elif statement
The elif keyword is Python's way of saying "if the previous conditions were not true, then try this condition

```python
# a is equal to b, so the first condition is not true, but the elif condition is true, so we print to screen that "a and b are equal"
x = 33
y = 33
if y > x:
  print("y is greater than x")
elif x == y:
  print("x and y are equal")
```

```python
# Multiple Elif Statements
score = 85

if score >= 90:
  print("Grade: A")
elif score >= 80:
  print("Grade: B")
elif score >= 70:
  print("Grade: C")
elif score >= 60:
  print("Grade: D")
```

### else statement
The else keyword catches anything which isn't caught by the preceding conditions.

The else statement is executed when the if condition (and any elif conditions) evaluate to False

```python
# x is greater than y, so the first condition is not true, so we go to the else condition and print to screen that "x is greater than y".
x = 200
y = 33
if y > x:
  print("y is greater than x")
else:
  print("x is greater than y")
```

```python
# Short Hand if-else
marks = 60
res = "Pass" if marks >= 40 else "Fail"
print(f"Result: {res}")
```

### Nested If Statements
You can have if statements inside if statements. This is called nested if statements.

```python
# In this example, the inner if statement only runs if the outer condition (age >= 18) is true.
age = 25
has_license = True

if age >= 18:
  if has_license:
    print("You can drive")
  else:
    print("You need a license")
else:
  print("You are too young to drive")
```

### pass statement
if statements cannot be empty, but if you for some reason have an if statement with no content, put in the pass statement to avoid getting an error.

```python
age = 16

if age < 18:
  pass # TODO: Add underage logic later
else:
  print("Access granted")
```
