### Functions
A function is a block of code which only runs when it is called,The idea is to put some commonly or repeatedly done task together and make a function so that instead of writing the same code again and again for different inputs.


```python
# a function is defined using the def keyword, followed by a function name and parentheses
def square(x):   # function name square
    return x**2
square(2)  # Calling a Function
```

```python
# We will create a function to check whether the number passed as an argument to the function is even or odd.
def evenOdd(x):
    if (x % 2 == 0):
        return "Even"
    else:
        return "Odd"

print(evenOdd(3))
print(evenOdd(8))
```

### Function Arguments
     Default Arguments
     Keyword Arguments
     Positional Arguments
     Arbitrary Arguments
     
     


```python
# Default Arguments: If the function is called without an argument, it uses the default value
def greet(name="Guest"):
    print(f"Hello {name}")

greet("Nouha")   # Hello Nouha
greet()           # Hello Guest


```

```python
# Keyword Arguments: You can send arguments with the key = value syntax, the order of the arguments does not matter.
def student(name, age):
    print(f"{name} is {age} years old")

student(name="Naya", age=15)
student(age=17, name="Sara")

```

```python
# Positional Arguments: When you call a function with arguments without using keywords,the order of the arguments does matter
def fav_animal(animal, name):
  print("I have a", animal)
  print("My", animal + "'s name is", name)

fav_animal("dog", "Buddy")
print("********")
fav_animal("kity","cat" )
```

```python
#Arbitrary Arguments: allow a function to accept a variable number of inputs. This is done using two special symbols:
# The *args parameter allows a function to accept any number of positional arguments.
def total(*numbers):
    print(sum(numbers))

total(1, 2, 3)        # 6
total(5, 10, 15, 20)  # 50
```

```python
#Arbitrary Arguments: 
# **kwargs: the function will receive a dictionary of arguments and can access the items accordingly
def info(**data):
    for key, value in data.items():
        print(f"{key}: {value}")

info(name="Naya", age=25, city="Amsterdam")
```

### lambda expressions
Lambda functions are small anonymous functions, meaning they do not have a defined name. These are small, short-lived functions used to pass simple logic to another function.

```python
add = lambda a : a + 10
print(add(5)) 
```

### map and filter


```python
# map() function in Python applies a given function to each element of an iterable (list, tuple, set, etc.)
# and returns a map object (iterator)
numbers = [1, 2, 3, 4]

squared = map(lambda x: x**2, numbers)

print(list(squared)) # [1, 4, 9, 16]
```

```python
# filter() function is used to extract elements from an iterable (like a list, tuple or set) that satisfy a given condition
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = filter(lambda x: x % 2 == 0, numbers)

print(list(even_numbers))  # [2, 4, 6]
```
